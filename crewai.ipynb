```python
import os
import gradio as gr

!pip install crewai
from crewai import Agent, Task, Crew, LLM

# ─── LLM Setup ──────────────────────────────────────────────────────────────

def build_llm(api_key: str, temperature: float = 0.7):
    os.environ["GOOGLE_API_KEY"] = api_key
    return LLM(model="gemini/gemini-2.5-flash", temperature=temperature)


# ─── Crew Builder ────────────────────────────────────────────────────────────

def run_crew(api_key, topic, temperature, progress=gr.Progress()):
    if not api_key.strip():
        raise gr.Error("Please enter your Google API Key.")

    progress(0.05, desc="Initialising LLM…")
    gemini_llm = build_llm(api_key.strip(), temperature)

    research_agent = Agent(
        role="AI Researcher",
        goal=f"Research latest trends in {topic}",
        backstory="Senior AI researcher expert in LLMs & Agentic AI",
        llm=gemini_llm,
        verbose=False,
    )

    writer_agent = Agent(
        role="Technical Writer",
        goal="Convert research into clear explanation",
        backstory="Expert in simplifying AI concepts",
        llm=gemini_llm,
        verbose=False,
    )

    reviewer_agent = Agent(
        role="AI Reviewer",
        goal="Ensure accuracy & clarity",
        backstory="Ensures content is workshop-ready",
        llm=gemini_llm,
        verbose=False,
    )

    research_task = Task(
        description=f"Research trends in {topic} (LLMs, Agents, Enterprise, Tools)",
        expected_output="Bullet summary",
        agent=research_agent,
    )

    writing_task = Task(
        description="Write structured technical explanation",
        expected_output="Article",
        agent=writer_agent,
    )

    review_task = Task(
        description="Review and polish content",
        expected_output="Final content",
        agent=reviewer_agent,
    )

    crew = Crew(
        agents=[research_agent, writer_agent, reviewer_agent],
        tasks=[research_task, writing_task, review_task],
        verbose=False,
    )

    result = crew.kickoff()
    return str(result)


# ─── LIGHT THEME CSS ─────────────────────────────────────────────────────────

CSS = """
body, .gradio-container {
    background: #f8fafc !important;
    color: #0f172a !important;
    font-family: 'Inter', sans-serif !important;
}

/* Header */
.header {
    text-align: center;
    padding: 2rem;
    border-bottom: 1px solid #e2e8f0;
}

.header h1 {
    font-size: 2.5rem;
    font-weight: 800;
    color: #2563eb;
    margin-bottom: 0.3rem;
}

.header p {
    color: #64748b;
    font-size: 0.9rem;
}

/* Panels */
.panel {
    background: #ffffff;
    border: 1px solid #e2e8f0;
    border-radius: 12px;
    padding: 1.2rem;
}

/* Inputs */
input, textarea {
    background: #ffffff !important;
    border: 1px solid #cbd5e1 !important;
    color: #0f172a !important;
}

/* Button */
#run-btn {
    background: #2563eb !important;
    color: white !important;
    border-radius: 8px !important;
    font-weight: bold !important;
}

/* Output */
#output-box textarea {
    min-height: 400px !important;
}
"""


# ─── UI ──────────────────────────────────────────────────────────────────────

with gr.Blocks(css=CSS, title="CrewAI") as demo:

    gr.HTML("""
    <div class="header">
        <h1>CrewAI</h1>
        <p>Multi-Agent Generative AI Research Studio</p>
    </div>
    """)

    with gr.Row():
        with gr.Column(scale=1, elem_classes="panel"):

            api_key = gr.Textbox(label="Google API Key", type="password")
            topic = gr.Textbox(label="Topic", value="Generative AI")
            temperature = gr.Slider(0, 1, value=0.7)

            run_btn = gr.Button("Run Crew", elem_id="run-btn")

        with gr.Column(scale=2, elem_classes="panel"):

            output = gr.Textbox(
                label="Output",
                lines=25,
                interactive=False,
                show_copy_button=True,
                elem_id="output-box",
            )

    run_btn.click(
        fn=run_crew,
        inputs=[api_key, topic, temperature],
        outputs=output,
    )


if __name__ == "__main__":
    demo.launch()
```
